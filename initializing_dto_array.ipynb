Initializing Different Types of Arrays

```python
import numpy as np
```

```python
#Initializing all 0s matrix
np.zeros((3,4),dtype=int) #3x4 matrix filled with 0s
```

```python
#Initializing all 1s matrix
np.ones((2,4)) #2x4 matrix filled with 1s
```

```python
#Initializing any other number matrix or random number matrix
np.full((2,2), 7,dtype=float) # 2x2 matrix filled with 7
```

```python
np.ones((2,4,3),dtype=int) #2x4x3 matrix filled with 1s ysema 3 chai harek element ko value vanamna
```
