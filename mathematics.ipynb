MATHEMATICS:

```python
import numpy as np
```

```python
a=np.array([1,2,3,4])
print(a)
```

```python
#Basic mathematical operations on arrays

print(a+1) #add 1 to each element
print(a-1) #subtract 1 from each element
print(a*2) #multiply each element by 2
print(a/2) #divide each element by 2
print(a**2) #square each element
```

```python
b=np.array([1,0,1,0])
print(a+b) #add two arrays
```

```python
#Calculating sign and cos values of an array
print(np.sin(a))
print(np.cos(a))
```

```python
#Acting upon individual elements of an array
a[1]=a[1]+1
print(a)
#Similarly we can perform other operations(-,*,/) on individual elements of an array.
```

    LINEAR ALGEBRA:

```python
#Matrix multiplication of two arrays
c=np.ones((2,3))
d=np.full((3,2),3)
print(np.matmul(c,d)) #matrix multiplication of two arrays
```

```python
#Matrix multiplication of two arrays
p=np.array([[1,2,3],[4,5,6]])
q=np.array([[7,8,],[10,11],[13,14]])
print(np.matmul(p,q))

'''Note:For matrix multiplication, the number of columns in the 
first array must be equal to the number of rows in the second array'''
```

```python
#Inverse of a matrix
o=np.linalg.inv(np.array([[1,2],[3,4]])) #inverse of a matrix
print(o)
```

```python
#Determinant of a matrix
det=np.linalg.det(np.array([[1,2],[3,4]])) #determinant of a matrix
print(det)
```

```python
#Addition of two matrices
a=np.zeros((3,3))
b=np.ones((3,3))
print(np.add(a,b)) #add two matrices
```

Statistics:

```python
stats=np.array([[1,2,3],[4,5,6]],dtype=int)
print(stats)
```

```python
#Basics:
print(np.min(stats,axis=0)) #minimum of each column
print(np.min(stats,axis=1)) #minimum of each row
print(np.min(stats))#minimum of the whole array
print(np.max(stats,axis=0)) #maximum of each column
print(np.max(stats,axis=1)) #maximum of each row
print(np.max(stats))#maximum of the whole array
```

Reorganizing Arrays

```python
before=np.array([[1,2,3,4],[3,4,5,6]]) #2*4 
print(before)
after=before.reshape((1,8)) #1*8
print(after)

#Similarly we can reshape the array to 4*2, 8*1, 2*2*2 etc.
#  but the total number of elements must be same.

```

```python
#Vertically stacking two arrays
a=np.array([[1,2,3,4]])
b=np.array([[5,6,7,8]])
print(np.vstack((a,b,b))) #vertically stacking two arrays
```

```python
#Horizontally stacking two arrays
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
print(np.hstack((b,a))) #horizontally stacking two arrays
```

**Miscellaneous:**

Load data from file

```python
#Loading data from a text file
data_file = np.genfromtxt('data.txt', delimiter=',',dtype=int)
print(data_file)
```

**Boolean Masking and Advanced Indexing**

```python
#Filtering data from a text file
data_file[data_file>80]
#data_file[data_file<80]
```

```python
#Checking if any element in each column is greater than 80
np.any(data_file>80,axis=0) #returns True if any element in the column is greater than 80
```

```python
#Filtering data from a text file based on multiple conditions
data_file[~((data_file>10) & (data_file<50))] #returns elements that are greater than 10 and less than 30
```

```python
(data_file>10) & (data_file<30) #returns a boolean array where True indicates the elements that are greater than 10 and less than 30
```
